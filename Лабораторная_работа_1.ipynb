Необходимо вычислить значение выражения:
$$F = a\_1+b - a\cdot(x + y5)
$$
Для каждой переменной необходимо вывести ее значение и тип.

## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $a\_1, b, a, x, y5$. Переменная, которая должна быть вычислена на основе независимых: $F$.

```python
a_1=20
b=10
a=5
x=3
y5=2
print("a_1 = ", a_1, ", type(a_1) = ", type(a_1))
print("b = ", b, ", type(b) = ", type(b))
print("a = ", a, ", type(a) = ", type(a))
print("x = ", x, ", type(x) = ", type(x))
print("y5 = ", y5, ", type(y5) = ", type(y5))
```

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $Z$:

```python
F= (a_1 + b) - a * (x + y5)
print('F=',F)
print('Type of F: ', type(F))
```
