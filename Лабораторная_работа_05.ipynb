## Задание

В файле pr5_1.txt содержится последовательность целых чисел. Элементы последовательности могут принимать целые значения от 0 до 10 000 включительно. Рассматривается множество элементов последовательности, которые оканчиваются на 5 или 7 и при этом не делятся ни на 9, ни на 11. Найдите количество таких чисел и сумму минимального и максимального из них. 

## Решение

```python
c = []
s = open('pr5_1.txt')
for i in s:
    i = int(i)
    if (str(i).endswith('5') or str(i).endswith('7')) and (i % 9 != 0 and i % 11 != 0):
        c += [i]
    else:
        continue
print(len(c), int(max(c)) + int(min(c)))
```
