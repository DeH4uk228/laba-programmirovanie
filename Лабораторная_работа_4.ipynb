## Задание

В файле data_v1.txt найдите самую длинную подпоследовательность не содержащую прописных букв и цифр, кратных 4. В ответе укажите саму подпоследовательность и ее длину. 


## Решение задачи 1

```python
f=open('data_v1.txt')
s=f.read().strip()

target=''
maxx=0
maxs=''
for c in s:
  if (c.isalpha() and c not in "ABCDEFGHIJKLMNOPQRSTUVWXYZ") or (c.isdigit() and int(c) % 4 != 0):
    target+=c
  else:
    if len(target)>maxx:
      maxx=len(target)
      maxs=target
      target=''

print('Максимальная длина строки = %d' % maxx)
print('Сама строка:', maxs)
```

1. Эти функции нужны для чтения файла, в этом их схожесть. Отличия только в методах: read() читает целиком, readline() - одну строку за вызов, а readlines() - возвращает все строки сразу, в виде списка. 

2. Многострочный текстовый файл можно прочитать, перебирая его строки с помощью цикла for, так как он поддерживает такой перебор.

3. Закрытие файла необходимо для прекращения работы с ним, а также сохранения как записанных на него новых данных, так и старых.

4. Позиция считывания файла - это место в файле, с которого программа начинает читать данные, или уже читает в данный момент. Первоначально она находится в 1 строке, перед 1 символом. 

5. Файл открывается в режиме дозаписи (a). Данные дописываются в конец

6. Исключение – это ошибка, которая происходит во время работы программы, приводящая к её внезапной остановке.

7. Если не обрабатывать исключение, и не препядствовать ему, то программа будет выдавать ршибку и аварийно завершаться. 

8. При попытке открыть несуществующий файл, программа выдаст исключение "FileNotFoundError" 

9. Программа вызовет исключение "ValueError", при попытке преобразовать нечисловую строку в float().
