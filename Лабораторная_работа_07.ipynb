## Задание

В файле содержатся сведения о сотрудниках компании, их зарплатах и 
сумме налога с зарплаты. 
1. Максимальная и минимальная з/п. Вывести соответствующих 
сотрудников. 
2. Средняя з/п сотрудников без указанного навыка (skill)? 
3. Количество сотрудников без e-mail. 
4. Суммарный налог. 

# Решение

### Подготовка Данных

```python
with open("p7_data_13.csv", "r", encoding="utf-8-sig") as file:
    lines = file.readlines()

headers = lines[0].strip().split(",")
employees = []

for line in lines[1:]:
    if line.strip() == "":
        continue

    values = line.strip().split(",")
    employee = {}

    for i in range(len(headers)):
        employee[headers[i].strip()] = values[i].strip()

    employee["salary_num"] = float(employee["salary"].strip("$"))
    employee["tax_num"] = float(employee["Tax"].strip("$"))

    employees.append(employee)

print("Всего сотрудников:", len(employees))
```

## Задание 1. Максимальная и минимальная зарплаты

```python
max_employee = employees[0]
min_employee = employees[0]

for employee in employees:
    if employee["salary_num"] > max_employee["salary_num"]:
        max_employee = employee
    if employee["salary_num"] < min_employee["salary_num"]:
        min_employee = employee

print(f"Максимальная зарплата: ${max_employee['salary_num']:.2f} — "
      f"{max_employee['FirstName']} {max_employee['LastName']}, ID: {max_employee['ID']}")
print(f"Минимальная зарплата: ${min_employee['salary_num']:.2f} — "
      f"{min_employee['FirstName']} {min_employee['LastName']}, ID: {min_employee['ID']}")
```

## Задание 2. Средняя зарплата без указанного навыка

```python
salary_sum = 0
no_skill_count = 0

for employee in employees:
    if employee["Skill"].strip() == "":
        salary_sum += employee["salary_num"]
        no_skill_count += 1

if no_skill_count > 0:
    average_salary = salary_sum / no_skill_count
    print("Сотрудников без указанного навыка:", no_skill_count)
    print(f"Средняя зарплата: ${average_salary:.2f}")
else:
    print("Сотрудников без указанного навыка нет.")
```

## Задание 3. Количество сотрудников без e-mail

```python
no_email_count = 0

for employee in employees:
    if employee["email"].strip() == "":
        no_email_count += 1

print("Сотрудников без e-mail:", no_email_count)
```

## Задание 4. Суммарный налог

```python
total_tax = 0

for employee in employees:
    total_tax += employee["tax_num"]

print(f"Суммарный налог: ${total_tax:.2f}")
```
