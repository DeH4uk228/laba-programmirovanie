# Уровень 2 - Средний

Задание: Пользователь вводит число n. С помощью цикла while разложите число на простые множители и выведите их в виде множитель1 * множитель2 * ....

~Пример: 60 = 2 * 2 * 3 * 5~


````python
n = int(input("Введите число: "))
a = n
otv = []
while n > 1:
    d = 2
    while n % d != 0:
        d += 1
    otv.append(str(d))
    n //= d
print(a, "=", ' * '.join(otv))
````

# Уровень 3 - Продвинутый

Задание: Пользователь вводит число n. Программа реализует угадывание числа от 1 до n методом бинарного поиска: компьютер загадывает, пользователь отвечает «больше»/«меньше»/«равно». Используйте while и input. Выведите количество попыток и сравните с теоретическим максимумом ⌈log2𝑛⌉ (используйте math.log2 и math.ceil).

````python
import math

n = int(input("Введите n: "))

low, high = 1, n
tries = 0

while low <= high:
    mid = (low + high) // 2
    tries += 1
    ans = input(f"Твоё число равно {mid}? (больше/меньше/равно): ").strip().lower()

    if ans == "равно":
        print(f"Угадал! Попыток: {tries}")
        break
    elif ans == "больше":      # загаданное больше mid
        low = mid + 1
    else:                      # "меньше" — загаданное меньше mid
        high = mid - 1

print(f"Теоретический максимум попыток: {math.ceil(math.log2(n))}")
````

# Контрольные Вопросы

## 1. Чем отличаются циклы `for` и `while`?

`for` используют, когда нужно перебрать элементы списка или выполнить действия заданное количество раз. `while` используют, когда повторение должно продолжаться, пока условие истинно, а число повторений заранее может быть неизвестно.

Например, `for` удобен для перебора оценок в списке, а `while` — для повторного запроса числа, пока пользователь не введёт ноль.

````python
# for: перебираем все числа в списке
ocenki = [4, 5, 3, 5]
for ocenka in ocenki:
    print(ocenka)

# while: повторяем действие, пока число не станет равным нулю
number = 3
while number != 0:
    print("Число ещё не равно нулю:", number)
    number -= 1
print("Цикл завершён")
````

## 2. Как работает `else` внутри цикла?

Блок `else` после `for` или `while` выполняется, если цикл завершился обычным способом — например, закончился диапазон или условие стало ложным. Если цикл прерван командой `break`, блок `else` **не выполняется**.

````python
# Пример 1: цикл завершился сам, поэтому else сработает
for number in range(3):
    print(number)
else:
    print("Цикл завершился без break")

# Пример 2: сработал break, поэтому else не выполнится
for number in range(5):
    if number == 2:
        break
    print(number)
else:
    print("Эта строка не появится")
````

## 3. Разница между `break` и `continue`

`break` полностью завершает ближайший цикл. `continue` пропускает оставшуюся часть текущей итерации и сразу переходит к следующей.

В примере ниже число 2 пропускается, на числе 5 цикл останавливается. Поэтому числа 6 и 7 уже не обрабатываются.

````python
number = 0
while number < 7:
    number += 1

    if number == 2:
        print("2 пропускаем с помощью continue")
        continue

    if number == 5:
        print("На 5 останавливаем цикл с помощью break")
        break

    print("Обрабатываем число", number)

# Выводятся числа 1, 3 и 4; для 2 и 5 показываются сообщения.
````

## 4. Типичные ошибки при организации циклов

**Ошибка 1 — бесконечный цикл:** переменная не меняется, поэтому условие всё время истинно.

```python
# Ошибка:
i = 0
while i < 5:
    print(i)  # i не увеличивается
```

**Исправление:** менять переменную внутри цикла.

**Ошибка 2 — неверный диапазон `range`:** верхняя граница не включается. `range(1, 5)` выдаёт числа от 1 до 4, а не до 5.

```python
# Если нужны числа от 1 до 5 включительно:
for i in range(1, 6):
    print(i)
```

**Ошибка 3 — условие, которого цикл не сможет дождаться:** если увеличивать число на 2, начиная с 1, оно никогда не станет равным 10.

```python
# Ошибка: условие i != 10 никогда не станет ложным
# i = 1
# while i != 10:
#     i += 2
```

**Исправление:** выбрать условие, которое действительно может завершить цикл, и шаг, соответствующий задаче.

````python
# Исправление ошибки 1: i увеличивается, цикл завершится
print("Ошибка 1 — исправленный вариант:")
i = 0
while i < 5:
    print(i)
    i += 1

# Исправление ошибки 3: проверяем границу, а не ожидание недостижимого числа
print("Ошибка 3 — исправленный вариант:")
i = 1
while i < 10:
    print(i)
    i += 2
# Напечатает 1, 3, 5, 7, 9, затем остановится.
````

## 5. Как работает `if / elif / else`?

`if` проверяет первое условие. Если оно ложно, Python проверяет `elif`. Если ни одно из условий не подошло, выполняется `else`. После первого истинного условия остальные ветви уже не проверяются.

Порядок важен, особенно когда условия пересекаются: слишком общее условие в начале может перехватить значения, предназначенные для следующей ветви.

````python
number = 7

if number < 0:
    print("Меньше 0")
elif 0 <= number <= 10:
    print("От 0 до 10 включительно")
else:
    print("Больше 10")

# Например, если первым написать if number >= 0,
# то числа больше 10 попадут в эту ветвь и до elif уже не дойдут.
````

## 6. Логические операторы `and`, `or`, `not`

- `and` — истинно, когда истинны **оба** условия.
- `or` — истинно, когда истинно **хотя бы одно** условие.
- `not` — меняет значение на противоположное: `True` на `False` и наоборот.

Условие «число чётное и от 10 до 20 включительно **либо** нечётное и больше 100» записывается одним выражением так:

````python
n = 14  # для примера; можно заменить другим числом
condition = ((n % 2 == 0 and 10 <= n <= 20) or (n % 2 != 0 and n > 100))
print(condition)
````

Здесь `% 2` даёт остаток от деления на 2: остаток 0 означает чётное число, а остаток 1 — нечётное. Скобки явно показывают, какие условия соединяются через `and`, а какие — через `or`.

## 7. Как вычисляются арифметические выражения?

Python соблюдает приоритет операций. В общем случае сначала выполняются действия в скобках, затем возведение в степень `**`, затем умножение, деление `/`, целочисленное деление `//` и остаток `%`, а после — сложение и вычитание. Операции одного уровня обычно выполняются слева направо; возведение в степень группируется справа налево.

Примеры:

````python
print(2 + 3 * 4)       # 14: сначала 3 * 4
print((2 + 3) * 4)     # 20: скобки выполняются первыми
print(17 // 5 + 17 % 5) # 5: 3 + 2
print(2 ** 3 ** 2)     # 512: это 2 ** (3 ** 2)
````

## 8. Чем отличаются `/` и `//`?

`/` выполняет обычное деление и возвращает результат типа `float` (например, `3.5`). `//` выполняет деление с округлением вниз до целого значения по числовой шкале. Если оба аргумента целые, результат обычно имеет тип `int`. При отрицательных числах округление идёт вниз, а не просто отбрасывает дробную часть.

````python
a = 7
b = 2

print(a / b)        # 3.5
print(a // b)       # 3
print(type(a / b))  # <class 'float'>
print(type(a // b)) # <class 'int'>

print(-7 // 2)      # -4, потому что результат округляется вниз
````

## 9. Встроенные функции `min`, `max`, `sum`, `abs`, `round`

- `min(...)` — возвращает наименьшее значение. Принимает несколько сравнимых значений или один итерируемый объект, например список.
- `max(...)` — возвращает наибольшее значение; аргументы устроены так же, как у `min`.
- `sum(...)` — складывает числовые элементы итерируемого объекта, например списка чисел.
- `abs(x)` — возвращает абсолютное значение числа, то есть расстояние от нуля без знака минус.
- `round(x, ndigits)` — округляет число. Второй аргумент необязателен и указывает количество знаков после запятой.

````python
numbers = [3, -1, 7, 2]

print(min(numbers))       # -1
print(max(numbers))       # 7
print(sum(numbers))       # 11
print(abs(-5))            # 5
print(round(3.14159, 2))  # 3.14
````

## 10. Как использовать `range()` в цикле `for`?

`range(start, stop, step)` создаёт последовательность целых чисел:

- `start` — первое число (по умолчанию 0);
- `stop` — граница, которая **не включается**;
- `step` — шаг (по умолчанию 1). Он может быть отрицательным для движения в обратном направлении и не должен быть равен нулю.

````python
print(list(range(1, 5)))       # шаг 1: [1, 2, 3, 4]
print(list(range(5, 0, -1)))   # отрицательный шаг: [5, 4, 3, 2, 1]
print(list(range(2, 15, 3)))   # произвольный шаг: [2, 5, 8, 11, 14]

# Чтобы получить числа от 1 до 5 включительно, пишут range(1, 6).
````

## 11. Как использовать модуль `math`?

Чтобы пользоваться функциями модуля, сначала импортируют его командой `import math`, затем указывают имя модуля: `math.sqrt(25)`.

Пять полезных функций:

- `math.sqrt(x)` — квадратный корень из неотрицательного числа;
- `math.ceil(x)` — округление вверх до ближайшего целого;
- `math.floor(x)` — округление вниз до ближайшего целого;
- `math.factorial(n)` — факториал целого неотрицательного числа;
- `math.sin(x)` — синус угла, заданного **в радианах**.

Частая ошибка — вызвать `sqrt(25)` после `import math`: такого имени отдельно нет, поэтому возникнет `NameError`. Нужно написать `math.sqrt(25)` либо импортировать функцию отдельно через `from math import sqrt`.

````python
import math

print(math.sqrt(25))       # 5.0
print(math.ceil(3.2))      # 4
print(math.floor(3.8))     # 3
print(math.factorial(5))   # 120
print(math.sin(math.pi/2)) # примерно 1.0
````

## 12. Три способа вычислить квадратный корень

**1. `math.sqrt(x)`** — стандартный и понятный способ для неотрицательных вещественных чисел. Для отрицательного `x` выдаёт `ValueError`.

**2. `x ** 0.5`** — короткая запись через возведение в степень. Для неотрицательных чисел обычно удобна, но при отрицательном числе в Python может получиться комплексное число, поэтому для обычного вещественного корня лучше заранее проверить аргумент.

**3. Метод Герона** — приближённый алгоритм: берут начальное приближение и много раз улучшают его. Он полезен, чтобы понять принцип численного вычисления, но требует больше кода, чем готовая функция.

````python
import math

x = 16
print(math.sqrt(x))  # 4.0
print(x ** 0.5)      # 4.0

def sqrt_geron(a, eps=1e-10):
    """Приближённо вычисляет корень неотрицательного числа методом Герона."""
    if a < 0:
        raise ValueError("Число должно быть неотрицательным")
    if a == 0:
        return 0.0

    guess = a if a >= 1 else 1.0
    while True:
        better_guess = (guess + a / guess) / 2
        if abs(better_guess - guess) < eps:
            return better_guess
        guess = better_guess

print(sqrt_geron(x))  # примерно 4.0
````

## 13. Как безопасно обрабатывать проблемные случаи?

Перед вычислением нужно проверить входные данные. Например, отрицательное число не имеет вещественного квадратного корня. Ошибку ввода (когда пользователь вводит не число) удобно обрабатывать через `try / except`.

````python
import math

try:
    number = float(input("Введите число для вычисления квадратного корня: "))
    if number < 0:
        print("Нельзя вычислить вещественный квадратный корень из отрицательного числа.")
    else:
        print("Квадратный корень:", math.sqrt(number))
except ValueError:
    print("Ошибка: пожалуйста, введите число, например 9 или 2.5.")
````

## 14. Почему результат `input()` нужно преобразовывать?

Функция `input()` всегда возвращает строку (`str`), даже если человек ввёл цифры. Поэтому выражение `'5' + 2` вызовет `TypeError`: строку нельзя напрямую сложить с целым числом. А выражение `'5' + '2'` даст текст `'52'`, а не число 7.

`int()` преобразует строку в целое число, а `float()` — в число с дробной частью. Если ввод не подходит, преобразование вызывает `ValueError`, который можно обработать через `try / except`.

````python
# Пример корректного преобразования с обработкой ошибки
try:
    number = float(input("Введите число: "))
    print("Число, умноженное на 2:", number * 2)
except ValueError:
    print("Ошибка: введено не число.")

# Для целого числа можно использовать int():
# age = int(input("Введите возраст: "))
````

## 15. Когда лучше цикл, а когда встроенные функции?

Допустим, нужно найти сумму чисел в списке. С помощью цикла можно накапливать сумму в переменной. С помощью `sum()` та же задача решается одной строкой.

Встроенная функция обычно короче и понятнее для стандартной задачи, а цикл удобен, когда при обработке каждого элемента нужны дополнительные проверки или действия.

````python
numbers = [3, 5, 2, 8]

# Вариант 1: цикл и накопление суммы
result = 0
for number in numbers:
    result += number
print("Сумма через цикл:", result)

# Вариант 2: встроенная функция
print("Сумма через sum():", sum(numbers))

# Сравнение:
# sum() — меньше кода, проще читать и труднее случайно ошибиться.
# Цикл — гибче, если для каждого числа нужны дополнительные действия.
````
