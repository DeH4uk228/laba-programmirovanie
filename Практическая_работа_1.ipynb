# Контрольные Вопросы

### 1. Что такое переменная в Python?

**Переменная** — это имя, которое используется для хранения значения. В Python значение присваивается переменной с помощью оператора `=`.

```python
age = 18
name = "Денис"
number = 3.14
```

При этом Python автоматически определяет тип значения.

**Корректные имена:**
```python
age = 18
student_name = "Иван"
number2 = 25
```

**Некорректное имя:**
```python
2number = 25
```

Оно недопустимо, потому что имя переменной **не может начинаться с цифры**. Также имя не должно содержать пробелы и специальные символы, а ключевые слова Python использовать нельзя.

### 2. Основные типы данных Python

Основные простейшие типы:

| Тип | Пример | `type()` |
|---|---|---|
| `int` — целое число | `25` | `<class 'int'>` |
| `float` — дробное число | `3.14` | `<class 'float'>` |
| `str` — строка | `"Hello"` | `<class 'str'>` |
| `bool` — логическое значение | `True` | `<class 'bool'>` |

Также существуют другие типы, например `list`, `tuple`, `dict`, `set`.

Тип переменной называется **динамическим**, потому что его не нужно заранее указывать. Одна и та же переменная может содержать значения разных типов:

```python
x = 10
x = "Hello"
x = 3.14
```

### 3. Как работает функция `print()`?

Функция `print()` используется для **вывода информации на экран**.

**1. Вывод одной переменной:**

```python
name = "Иван"
print(name)
```

Результат:

```text
Иван
```

**2. Вывод нескольких значений через запятую:**

```python
name = "Иван"
age = 18
print(name, age)
```

Результат:

```text
Иван 18
```

Python автоматически ставит пробел между значениями.

**3. Форматированное сообщение:**

```python
name = "Иван"
age = 18
print(f"Студента зовут {name}, ему {age} лет")
```

Результат:

```text
Студента зовут Иван, ему 18 лет
```

Разница в том, что при обычном выводе значения просто выводятся, а **f-строка позволяет вставлять переменные непосредственно внутрь текста**.

### 4. Что делает `type()`?

Функция `type()` определяет **тип объекта или переменной**.

```python
number = 25
text = "Hello"
flag = True

print(type(number))
print(type(text))
print(type(flag))
```

Результат:

```text
<class 'int'>
<class 'str'>
<class 'bool'>
```

Это означает:
- `number` — целое число (`int`);
- `text` — строка (`str`);
- `flag` — логическое значение (`bool`).

### 5. Арифметические операции в Python

Основные арифметические операции:

| Оператор | Операция | Пример | Результат | Тип |
|---|---|---|---|---|
| `+` | сложение | `5 + 2` | `7` | `int` |
| `-` | вычитание | `5 - 2` | `3` | `int` |
| `*` | умножение | `5 * 2` | `10` | `int` |
| `/` | обычное деление | `5 / 2` | `2.5` | `float` |
| `//` | целочисленное деление | `5 // 2` | `2` | `int` |
| `%` | остаток от деления | `5 % 2` | `1` | `int` |
| `**` | возведение в степень | `5 ** 2` | `25` | `int` |

Если используются `float`, результат некоторых операций также будет `float`:

```python
5.0 + 2       # 7.0
5.0 * 2       # 10.0
```

Особенность `/`: обычное деление **всегда возвращает `float`**:

```python
10 / 2
```

Результат:

```text
5.0
```

### 6. Разница между `/` и `//`

`/` — обычное деление, результатом которого является `float`.

`//` — целочисленное деление, при котором результат округляется **вниз**.

Для положительных чисел:

```python
7 / 2
# 3.5

7 // 2
# 3
```

```python
10 / 3
# 3.3333333333333335

10 // 3
# 3
```

Для отрицательных чисел важно, что `//` округляет именно **в меньшую сторону**, а не просто отбрасывает дробную часть:

```python
-7 / 2
# -3.5

-7 // 2
# -4
```

```python
-10 / 3
# -3.3333333333333335

-10 // 3
# -4
```

То есть `//` выбирает ближайшее целое **в сторону минус бесконечности**.

### 7. Преобразование типов

Для преобразования типов используются специальные функции.

**`int()` — преобразование в целое число:**

```python
x = int("25")
print(x)
# 25
```

**`float()` — преобразование в дробное число:**

```python
x = float("3.14")
print(x)
# 3.14
```

**`str()` — преобразование в строку:**

```python
x = str(25)
print(x)
# "25"
```

Ошибка может возникнуть, если преобразование невозможно:

```python
x = int("hello")
```

Возникнет ошибка `ValueError`, потому что строку `"hello"` нельзя преобразовать в число.

### 8. Вычисление в Python как выражение, возвращающее значение

**Выражение** — это комбинация значений и операторов, результатом которой является некоторое значение.

**Арифметическое выражение:**

```python
x = 5 + 3
print(x)
```

Результат — `8`.

**Сравнение:**

```python
x = 10 > 5
print(x)
```

Результат — `True`.

**Логическое выражение:**

```python
x = (10 > 5) and (3 < 7)
print(x)
```

Результат — `True`.

Таким образом, результат любого выражения можно сохранить в переменную или сразу передать в `print()`.

### 9. Операторы сравнения и логические операторы

Операторы сравнения:

- `==` — равно;
- `!=` — не равно;
- `<` — меньше;
- `>` — больше;
- `<=` — меньше или равно;
- `>=` — больше или равно.

Результатом сравнения всегда является `True` или `False`.

Логические операторы:

- `and` — И;
- `or` — ИЛИ;
- `not` — НЕ.

Примеры:

```python
print(5 > 3 and 10 > 7)
```

Результат:

```text
True
```

Оба условия истинны.

```python
print(5 > 10 or 7 > 3)
```

Результат:

```text
True
```

Хотя бы одно условие истинно.

```python
print(not (5 > 3))
```

Результат:

```text
False
```

`not` изменяет значение на противоположное.

**Приоритет** в общем случае такой:

1. скобки;
2. арифметические операции;
3. сравнения;
4. `not`;
5. `and`;
6. `or`.

### 10. Операции над операндами разных типов

Python иногда может автоматически привести типы.

Например, при операции `int` и `float`:

```python
x = 5 + 2.5
print(x)
```

Результат:

```text
7.5
```

`int` автоматически учитывается как `float`.

Но автоматическое преобразование работает не всегда.

Например:

```python
x = "5" + 2
```

вызовет ошибку `TypeError`, потому что Python не складывает строку и число автоматически.

Нужно выполнить явное преобразование:

```python
x = int("5") + 2
print(x)
```

Результат:

```text
7
```

### 12. Порядок выполнения операций

Python выполняет операции в определённом порядке. Скобки имеют наивысший приоритет и позволяют изменить этот порядок.

Например:

```python
result = (2 + 3) * 4 - 6 / 2
print(result)
```

Сначала выполняется выражение в скобках:

```text
2 + 3 = 5
```

Затем умножение и деление:

```text
5 * 4 = 20
6 / 2 = 3.0
```

После этого вычитание:

```text
20 - 3.0 = 17.0
```

Результат:

```text
17.0
```

### 13. Множественное присваивание

Множественное присваивание позволяет одновременно присвоить значения нескольким переменным:

```python
a, b, c = 10, 20, 30
```

Теперь:

```text
a = 10
b = 20
c = 30
```

Также можно **обменять значения двух переменных без дополнительной переменной**:

```python
a = 5
b = 10

a, b = b, a
```

Теперь `a = 10`, а `b = 5`.

Можно использовать распаковку списка или кортежа:

```python
numbers = [10, 20, 30]
a, b, c = numbers
```

В результате:

```text
a = 10
b = 20
c = 30
```

### 14. Изменяемые и неизменяемые типы

**Изменяемые типы** позволяют изменять содержимое существующего объекта.

**Неизменяемые** типы нельзя изменить после создания объекта. К ним относятся, например, числа (`int`, `float`) и строки (`str`).

Например, строку нельзя изменить напрямую:

```python
text = "Hello"
# text[0] = "h"
```

Такой код вызовет ошибку `TypeError`.

Можно создать **новую строку**:

```python
text = "Hello"
text = "hello"
```

Здесь старая строка не изменяется — создаётся новое значение, на которое начинает ссылаться переменная `text`.

Это можно увидеть с помощью `id()`:

```python
text = "Hello"
print(id(text))

text = "hello"
print(id(text))
```

Значения `id()` могут отличаться, потому что после присваивания переменная стала ссылаться на другой объект.

### 15. Небольшой фрагмент программы

```python
a = 10
b = 2.5
name = "Python"
result = a * b
print(name, result)
print(type(a), type(b), type(result))
```

**Пояснение:**

1. `a = 10` — создаётся переменная `a` со значением `10`, тип `int`.
2. `b = 2.5` — создаётся переменная `b`, тип `float`.
3. `name = "Python"` — создаётся строковая переменная, тип `str`.
4. `result = a * b` — выполняется умножение `10 * 2.5`, результат `25.0`, тип `float`.
5. `print(name, result)` — выводятся строка `"Python"` и число `25.0`.
6. `print(type(a), type(b), type(result))` — выводятся типы трёх переменных.

Результат:

```text
Python 25.0
<class 'int'> <class 'float'> <class 'float'>
```
