## Задание

Григорий придумывает 16-буквенные слова, состоящие из букв слова АНТИУТОПИЯ. Сколько слов, содержащих комбинацию АНТИУТОПИЯ, может составить Григорий, если справа от этой комбинации находится равное количество гласных и согласных, а слева – не больше 2-х согласных? Буквы в словах могут повторяться любое количество раз или же не встречаться вовсе. 

## Решение

```python
from math import comb

VOWELS, CONS = 'АИОУЯ', 'НТП'
COMB, N = 'НТИУТОПИЯ', 16

BLOCK = len(COMB)

c = 0
for left_len in range(N - BLOCK + 1):
    right_len = N - BLOCK - left_len
    if right_len % 2:
        continue
    left = sum(comb(left_len, k) * len(CONS)**k * len(VOWELS)**(left_len - k)
               for k in range(3))
    right = (comb(right_len, right_len // 2)
             * len(VOWELS)**(right_len // 2) * len(CONS)**(right_len // 2))
    c += left * right
print(c)
```

**5. Функции и классы модуля `itertools`:**

- **`combinations()`** — все сочетания из n элементов по k.
- **`combinations_with_replacement()`** — сочетания с повторениями.
- **`chain()`** — объединяет несколько последовательностей в одну.
- **`count()`** — бесконечный счётчик.
- **`cycle()`** — бесконечно повторяет последовательность.
- **`accumulate()`** — накопительные суммы (или другая операция).
